## A)

find radius and corner freq for Mw 5, 6, 7, 8, 9

```python
#fc=?
#r=?

beta = 3.5 # km/s
k = 0.32
deltasigma = 3e6 # 3 MPa -> N/m^2
Mw = np.array([5, 6, 7, 8, 9])

def A(Mw, beta, deltasigma, k):
  M0 = 10**((3/2) * Mw + 9.1) # N*m

  r_m = np.cbrt((7 * M0)/(16 * deltasigma)) # in meters

  # m -> km
  r_km = r_m / 1000.0

  fc = (k * beta)/r_km # Hz

  print(f"Mw = {Mw} -> r = {r_km:.3f} km,  fc = {fc:.5f} Hz")
  return r_km, fc, M0

#Mw 5-9
for i in Mw:
  A(i, beta, deltasigma, k)
```

## B)

At what Mw does fc = 1Hz (the mb measurement freq)? At what Mw does it equal 0.05 Hx (the 20s period used for Ms)

```python
def Mw_fc (fc, beta, deltasigma, k):
  r_km = (k * beta)/fc #in km

  r_m = r_km * 1000 # in m

  M0 = (16/7) * deltasigma * (r_m**3) # N*m

  # Added missing * operator and brackets to match Mw = (2/3) * (log10(M0) - 9.1)
  Mw = (2/3) * (np.log10(M0) - 9.1)

  print(f"fc = {fc:.5f} Hz -> Mw = {Mw:.2f}")
  return Mw

fc = np.array([1, 0.05])
for i in fc:
  Mw_fc(i, beta, deltasigma, k)
```

## C)

Define toy mag at freq f as m(f) = (2/3)lod10A(f) + const, which equals Mw when f<<fc. Compute m(1 Hz) for Mw 5-9, and the increase in m for each one-unit increase in Mw

```python
import numpy as np

# Input parameters from Part A
beta = 3.5 # km/s
k = 0.32
deltasigma = 3e6 # Pa
Mw_values = np.array([5, 6, 7, 8, 9])
f_targets = [1.0, 0.05] # Hz
const = -(2/3) * 9.1

for f_target in f_targets:
    print(f"f = {f_target} Hz")
    m_list = []
    for Mw in Mw_values:
        M0 = 10**((3/2) * Mw + 9.1)

        r_meters = np.cbrt((7 * M0) / (16 * deltasigma))
        r_km = r_meters / 1000.0
        fc = (k * beta) / r_km

        Af = M0 / (1 + (f_target / fc)**2)
        mf = (2/3) * np.log10(Af) + const
        m_list.append(mf)

        print(f"Mw = {Mw} -> fc = {fc:.4f} Hz | m({f_target} Hz) = {mf:.2f}")

    print(f"\n m({f_target} Hz) increase for 1-unit increase in Mw:")
    for i in range(len(m_list) - 1):
        diff = m_list[i+1] - m_list[i]
        print(f"  Mw {Mw_values[i]} -> {Mw_values[i+1]}: increase = {diff:.3f}")
    print(f"\n")
```

## D)

2011 Tohoku eq has Mw 9.1 but mb=7.2. Why can GNSS peak ground displace keep growing? Using log10(PGD)=-3.841 + 0.937 Mw - 0.127 Mw log10R, compare PGD at R=300km for Mw 8 and 9.

Traditional seismic magnitudes are measured at high fixed frequencies. So, measuring for a giant earhtquake like Tohoku on captures a tiny oversaturated portion of the energy spectrum. \
In the meantime, GNSS measures the permanent deformation and extremely long period wavefields. So it can measure lower frequencies, and the displacement is proportional to the sesimic moment.

```python
#compare PGD R=300km for Mw 8&9
Mw = np.array([8, 9])
R = 300 #km
for i in Mw:
  PGD = 10**(-3.841 + 0.937*i - (0.127 * i * (np.log10(R))))
  print(f"Mw = {i} -> PGD = {PGD:.2f} m")
```
